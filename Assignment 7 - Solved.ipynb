#### Employee Data Bank.



Make a class called Employee. The __init__() method for Employee should store its different attributes: name, number, address, salary, post, year of joining, education, and his year of experience.

Make a method called describe_employee() that prints the detail Information about employee, a method update_ year_of_experience(), and a method called increase_salary() that do increamnet in it's salary.

Make an instance called employee101 from your class. Print the four attri-
butes individually, and then call all it's methods for describing employee, ssalary increamnet and increase year of experience.

Create 10 different instances from the class, and call describe_employee() for each
instance. Then do some updations with it's methods.



```python
class Employee:
    def __init__(self, name, number, address, salary, post, year_of_joining, education, years_of_experience):
        self.name = name
        self.number = number
        self.address = address
        self.salary = salary
        self.post = post
        self.year_of_joining = year_of_joining
        self.education = education
        self.years_of_experience = years_of_experience

    def describe_employee(self):
        print("Name:", self.name)
        print("Number:", self.number)
        print("Address:", self.address)
        print("Salary:", round(self.salary, 2))
        print("Post:", self.post)
        print("Year of Joining:", self.year_of_joining)
        print("Education:", self.education)
        print("Years of Experience:", self.years_of_experience)
        print("-" * 40)

    def update_year_of_experience(self, years=1):
        self.years_of_experience += years
        print(f"{self.name}'s experience updated to {self.years_of_experience} years")

    def increase_salary(self, percent):
        self.salary += self.salary * percent / 100
        print(f"{self.name}'s salary increased by {percent}% to {round(self.salary, 2)}")
```

```python
employee101 = Employee("Ali Khan", 101, "Karachi", 80000, "Software Engineer", 2020, "BS Computer Science", 5)

print(employee101.name)
print(employee101.number)
print(employee101.post)
print(employee101.salary)
print()

employee101.describe_employee()
employee101.increase_salary(10)
employee101.update_year_of_experience()
employee101.describe_employee()
```

```python
def create_employee(name, number, address, salary, post, year_of_joining, education, years_of_experience):
    return {
        "name": name,
        "number": number,
        "address": address,
        "salary": salary,
        "post": post,
        "year_of_joining": year_of_joining,
        "education": education,
        "years_of_experience": years_of_experience,
    }


def describe_employee(employee):
    print("Name:", employee["name"])
    print("Number:", employee["number"])
    print("Address:", employee["address"])
    print("Salary:", round(employee["salary"], 2))
    print("Post:", employee["post"])
    print("Year of Joining:", employee["year_of_joining"])
    print("Education:", employee["education"])
    print("Years of Experience:", employee["years_of_experience"])
    print("-" * 40)


def update_year_of_experience(employee, years=1):
    employee["years_of_experience"] += years
    print(f"{employee['name']}'s experience updated to {employee['years_of_experience']} years")


def increase_salary(employee, percent):
    employee["salary"] += employee["salary"] * percent / 100
    print(f"{employee['name']}'s salary increased by {percent}% to {round(employee['salary'], 2)}")
```

```python
employee101 = create_employee("Ali Khan", 101, "Karachi", 80000, "Software Engineer", 2020, "BS Computer Science", 5)

print(employee101["name"])
print(employee101["number"])
print(employee101["post"])
print(employee101["salary"])
print()

describe_employee(employee101)
increase_salary(employee101, 10)
update_year_of_experience(employee101)
describe_employee(employee101)
```
